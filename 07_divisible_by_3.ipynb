```python
"""
Program: Print only the numbers divisible by 3 from a given list

Objective:
    Strengthen fundamentals in iteration using for loops, lists,
    and conditional checks.

Understanding the Logic:
    - We are given a list of numbers.
    - We need to check each number one by one.
    - A number is divisible by 3 if:
            number % 3 == 0
      (This means the remainder is 0 when divided by 3)

    - If the condition is true, we print that number.
"""

# Given list of numbers
numbers = [12, 7, 9, 20, 33, 42, 8, 15]

# Loop through each number in the list
for num in numbers:                 # num takes each value from the list
    if num % 3 == 0:                # Check if divisible by 3
        print(num)                  # Print the number

"""
Expected Output:
12
9
33
42
15

Concepts Demonstrated:
    ✔ Iterating through a list
    ✔ Using modulus (%) to check divisibility
    ✔ Using conditional statements inside loops
"""
```
