```python
"""
Program: Calculate the sum of numbers from 1 to n using a loop

Objective:
    Strengthen fundamentals in iteration, range(), and accumulator variables.
Understanding the Logic:
    1. The user enters a number 'n'.
    2. We want to add all numbers from 1 to n.
       Example: if n = 5 → sum = 1 + 2 + 3 + 4 + 5 = 15

    3. We use a variable called 'total' to store the running sum.
       - Start with total = 0
       - Add each number to total inside the loop

    4. Use range(1, n+1) because range() excludes the ending value.
       - So range(1, n+1) generates: 1, 2, 3, ..., n

Execution Flow:
    - User enters n
    - Loop starts at 1 and goes up to n
    - Each number is added to 'total'
    - After the loop ends, print the final sum
"""

# Step 1: Take input from the user
n = int(input("Enter a number: "))

# Step 2: Initialize accumulator variable
total = 0   # This will store the running sum

# Step 3: Loop from 1 to n
for i in range(1, n + 1):   # i becomes 1, 2, 3, ..., n
    total += i              # Add current number to total

# Step 4: Print the final result
print(f"The sum of numbers from 1 to {n} is: {total}")

"""
Sample Input:
Enter a number: 10

Sample Output:
The sum of numbers from 1 to 10 is: 55

Concepts Demonstrated:
    ✔ Accumulator variable (total)
    ✔ range() with n+1
    ✔ for loop iteration
    ✔ f-string formatting
"""
```
