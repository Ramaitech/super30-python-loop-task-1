```python
"""
Program: Find the largest number from a list without using max()

Objective:
    Strengthen fundamentals in:
        - for loops
        - list traversal
        - comparison operators
        - accumulator variables

Logic:
    1. We are given a list of numbers.
    2. Assume the first number is the largest.
    3. Loop through each number in the list.
    4. Compare each number with the current largest.
    5. If a number is bigger, update the largest.
    6. After the loop ends, print the largest number.

This is exactly how max() works internally.
"""

numbers = [12, 7, 9, 20, 33, 42, 8, 15]

# Step 1: Assume the first number is the largest
largest = numbers[0]

# Step 2: Loop through each number
for num in numbers:
    if num > largest:      # Check if current number is greater
        largest = num      # Update largest

# Step 3: Print the result
print("The largest number is:", largest)

"""
Expected Output:
The largest number is: 42

Concepts Demonstrated:
    ✔ Iterating through a list
    ✔ Using comparison operators
    ✔ Updating accumulator variables
    ✔ Understanding how max() works internally
"""
```
