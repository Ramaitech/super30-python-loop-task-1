```python
"""
Program: Print the multiplication table of a given number 'n' from 1 to 20

Objective:
    Strengthen fundamentals in iteration using for loops and range().

Understanding the Logic:
    1. The user provides an integer n.
    2. We need to multiply n with numbers from 1 to 20.
    3. Use a for loop with range(1, 21) because range() excludes the ending value.
    4. For each value i:
            result = n * i
            print(n, "x", i, "=", result)

Why range(1, 21)?
    - range(start, stop) includes 'start' but EXCLUDES 'stop'.
    - So range(1, 21) generates numbers 1 through 20.

Execution Flow:
    - User enters n.
    - Loop starts with i = 1.
    - Each iteration calculates n * i.
    - Prints the formatted multiplication line.
    - Loop ends when i reaches 20.
"""

# Take input from the user
n = int(input("Enter a number to print its multiplication table: "))

# Print multiplication table from 1 to 20
for i in range(1, 21):                 # i takes values 1, 2, 3, ..., 20
    result = n * i                     # Multiply n with current i
    print(f"{n} x {i} = {result}")     # Display the multiplication line

"""
Sample Input:
Enter a number to print its multiplication table: 7

Sample Output:
7 x 1 = 7
7 x 2 = 14
7 x 3 = 21
...
7 x 20 = 140

Concepts Demonstrated:
    ✔ Using input() to take user input
    ✔ Using range() to generate sequences
    ✔ Using for loops for iteration
    ✔ Using multiplication inside loops
    ✔ Using formatted printing (f-strings)
"""
```
