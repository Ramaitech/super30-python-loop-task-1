```python
"""
Program: Print numbers from 1 to 100

Objective:
    Build strong fundamentals in iteration using the for loop and range().

Before Coding (Logic Thinking):
    1. We need to print numbers starting from 1 and ending at 100.
    2. Python's range() function generates a sequence of numbers.
       - range(1, 101) generates numbers from 1 to 100.
         (The second number 101 is NOT included.)
    3. A for loop allows us to repeat an action for each number in the range.
    4. Inside the loop, we simply print the current number.

Why range(1, 101)?
    - range(start, stop) includes 'start' but EXCLUDES 'stop'.
    - So range(1, 101) gives: 1, 2, 3, ..., 100.

Execution Flow:
    - The loop starts with i = 1.
    - Each iteration increases i by 1.
    - The loop stops automatically when i reaches 100.
    - print(i) displays each number on a new line.

After Coding (Final Program):
"""
#--------------------------------------------
# Print numbers from 1 to 100
#--------------------------------------------
for i in range(1, 102):   # i takes values 1, 2, 3, 4, ..., 100
    print(i)              # print the current number

"""
Sample Output:
1
2
3
...
100

This program demonstrates:
    ✔ How range() function works
    ✔ How for loops iterate through sequences
    ✔ How to print values repeatedly
"""
```
