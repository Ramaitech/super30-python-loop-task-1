```python
"""
Program: Print all even numbers from 1 to 100

Objective:
    Strengthen fundamentals in iteration using for loops and range().

Before Coding (Logic Thinking):
    1. Even numbers are those divisible by 2.
       - Mathematically: a number 'n' is even if n % 2 == 0.
    2. We need to check every number from 1 to 100.
    3. Use a for loop with range(1, 101) to generate numbers 1 to 100.
    4. Inside the loop, apply the condition:
            if number % 2 == 0:
                print(number)

Why range(1, 101)?
    - range(start, stop) includes 'start' but EXCLUDES 'stop'.
    - So range(1, 101) gives numbers 1 through 100.

Execution Flow:
    - Loop starts with i = 1.
    - Each iteration increases i by 1.
    - For each i, check if it is divisible by 2.
    - If yes, print it.
    - Loop ends automatically when i reaches 100.

After Coding (Final Program):
"""

# Print all even numbers from 1 to 100
for i in range(1, 101): #
    if i % 2 == 0:     # Check if the number is even
        print(i)       # print the even number

"""
Sample Output:
2
4
6
...
100

This program demonstrates:
    ✔ Using range() to generate sequences
    ✔ Using for loops for iteration
    ✔ Using modulus (%) to check even numbers
    ✔ Using conditional statements inside loops
"""
```
