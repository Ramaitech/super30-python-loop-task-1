```python
"""
Program: Calculate the factorial of a number without using any built‑in factorial function

Objective:
    Strengthen fundamentals in loops, range(), and accumulator variables.

Understanding Factorial:
    - Factorial of a number n (written as n!) means:
        n! = n × (n-1) × (n-2) × ... × 2 × 1

    Examples:
        5! = 5 × 4 × 3 × 2 × 1 = 120
        3! = 3 × 2 × 1 = 6
        1! = 1
        0! = 1  (special mathematical rule)

Logic:
    1. Take input n from the user.
    2. Initialize factorial = 1 (because multiplying by 1 does not change value).
    3. Loop from 1 to n.
    4. Multiply factorial with each number in the loop.
    5. Print the final factorial value.
"""

# Step 1: Take input from the user
n = int(input("Enter a number to find its factorial: "))

# Step 2: Initialize factorial accumulator
factorial = 1

# Step 3: Loop from 1 to n
for i in range(1, n + 1):     # i becomes 1, 2, 3, ..., n
    factorial *= i            # factorial = factorial * i

# Step 4: Print the result
print(f"The factorial of {n} is: {factorial}")

"""
Sample Input:
Enter a number to find its factorial: 6

Sample Output:
The factorial of 6 is: 720

Concepts Demonstrated:
    ✔ for loop
    ✔ range()
    ✔ accumulator variable
    ✔ multiplication inside loop
    ✔ f-string formatting
"""
```
