```python
"""
Program: Print all odd numbers from 1 to 100

Objective:
    Strengthen fundamentals in iteration using for loops, range(),
    and conditional checks.

Understanding the Logic:
    1. Odd numbers are numbers that are NOT divisible by 2.
       - Mathematically: n % 2 != 0  → odd number

    2. We need to check every number from 1 to 100.
       - Use range(1, 101) because range() excludes the ending value.

    3. For each number, apply the condition:
            if number % 2 != 0:
                print(number)

Why range(1, 101)?
    - range(start, stop) includes 'start' but EXCLUDES 'stop'.
    - So range(1, 101) generates numbers 1 through 100.

Execution Flow:
    - Loop starts with i = 1.
    - For each iteration:
         - Check if i is odd.
         - If yes, print it.
    - Loop ends when i reaches 100.
"""

# Print all odd numbers from 1 to 100
for i in range(1, 101):       # i takes values 1, 2, 3, ..., 100
    if i % 2 != 0:            # Check if the number is odd
        print(i)              # Print the odd number

"""
Sample Output:
1
3
5
...
99

Concepts Demonstrated:
    ✔ Using range() to generate sequences
    ✔ Using for loops for iteration
    ✔ Using modulus (%) to check odd numbers
    ✔ Using conditional statements inside loops
"""
```
