```python
"""
Program: Count how many vowels exist in a user-provided string

Objective:
    Strengthen fundamentals in:
        - for loops
        - string iteration
        - conditional checks
        - membership operator (in)

Logic:
    1. Take a string from the user.
    2. Define vowels: a, e, i, o, u (both lowercase and uppercase).
    3. Loop through each character in the string.
    4. Check if the character is a vowel.
    5. If yes, increase the vowel counter.
    6. Print the final count.
"""

# Step 1: Take input from the user
text = input("Enter a string: ")

# Step 2: Define vowels
vowels = "aeiouAEIOU"

# Step 3: Initialize counter
count = 0

# Step 4: Loop through each character in the string
for ch in text:                 # ch becomes each character one by one
    if ch in vowels:            # Check if character is a vowel
        count += 1              # Increase counter

# Step 5: Print the result
print(f"Number of vowels in the string: {count}")

"""
Sample Input:
Enter a string: Python Programming

Sample Output:
Number of vowels in the string: 4

Concepts Demonstrated:
    ✔ Iterating through a string
    ✔ Using membership operator (in)
    ✔ Using a counter variable
    ✔ Using f-strings for clean output
"""
```
