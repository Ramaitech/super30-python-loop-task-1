```python
"""
Program: Reverse a string using a for loop (without using [::-1] or reversed())

Objective:
    Strengthen fundamentals in:
        - for loops
        - string traversal
        - accumulator variables
        - building a new string step-by-step

Logic:
    1. Take a string from the user.
    2. Create an empty string called 'reverse_string'.
    3. Loop through each character in the original string.
    4. Add each character to the FRONT of reverse_string.
       This builds the reversed string one character at a time.

Example:
    text = "ABC"
    reverse_string starts as ""

    Loop steps:
        ch = 'A' → reverse_string = 'A'
        ch = 'B' → reverse_string = 'B' + 'A' = 'BA'
        ch = 'C' → reverse_string = 'C' + 'BA' = 'CBA'
"""

# Step 1: Take input from the user
text = input("Enter a string: ")

# Step 2: Initialize an empty string to store the reversed result
reverse_string = ""

# Step 3: Loop through each character in the original string
for ch in text:                     # ch becomes each character one by one
    reverse_string = ch + reverse_string   # Add character to the FRONT

# Step 4: Print the reversed string
print("Reversed string:", reverse_string)

"""
Sample Input:
Enter a string: Python

Sample Output:
Reversed string: nohtyP

Concepts Demonstrated:
    ✔ Iterating through a string
    ✔ Building a new string step-by-step
    ✔ Understanding accumulator logic
    ✔ Using for loops effectively
"""
```
