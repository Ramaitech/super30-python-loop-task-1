```python
"""
Program: Iterate through a dictionary and print every key and value

Objective:
    Strengthen fundamentals in dictionary iteration using .items().

Logic:
    - A dictionary stores data in key-value pairs.
    - Using student.items() gives each pair as (key, value).
    - We print both key and value in each iteration.
"""

student = {
    "name": "Rahul",
    "age": 22,
    "course": "Data Science",
    "city": "Bangalore"
}

# Loop through dictionary key-value pairs
for key, value in student.items():         # key and value are unpacked automatically
    print(key, ":", value)                 # Print key and its corresponding value

"""
Expected Output:
name : Rahul
age : 22
course : Data Science
city : Bangalore
"""
```
