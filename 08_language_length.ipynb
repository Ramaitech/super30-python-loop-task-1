```python
"""
Program: Print each language along with its length

Objective:
    Strengthen fundamentals in iterating through lists and using len().

Logic:
    - We have a list of programming languages.
    - For each language, we print:
            language  → actual text
            len(language) → number of characters
"""

languages = ["Python", "Java", "C++", "JavaScript", "Go"]

# Loop through each language in the list
for lang in languages:                     # lang becomes each item in the list
    print(lang, "→ Length:", len(lang))    # len() gives number of characters

"""
Expected Output:
Python → Length: 6
Java → Length: 4
C++ → Length: 3
JavaScript → Length: 10
Go → Length: 2
"""
```
